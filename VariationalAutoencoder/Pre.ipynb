# 似然函数（likelihood function）与密度函数（density function / probability density function, PDF）？


---

## 核心区别一句话理解

> **密度函数（Density Function）** 是从“参数生成数据”的角度建模；
> **似然函数（Likelihood Function）** 是从“数据推断参数”的角度建模。

它们常常**使用同一个数学表达式**，但**视角不同、变量不同、用途不同**。

---

## 1️定义区分

| 名称                   | 数学符号                                  | 变量视角                       | 类型     | 含义                                  |
| -------------------- | ------------------------------------- | -------------------------- | ------ | ----------------------------------- |
| **密度函数（PDF）**        | $p(x \mid \theta)$                    | \$x\$ 为变量，\$\theta\$ 为常量   | 概率密度函数 | 在参数 \$\theta\$ 下，变量 \$x\$ 服从的分布     |
| **似然函数（Likelihood）** | $L(\theta \mid x) = p(x \mid \theta)$ | \$\theta\$ 为变量，\$x\$ 为已知数据 | 函数     | 在观测到 \$x\$ 后，参数 \$\theta\$ 的可能性大小度量 |

> 相同公式，不同解释：
>
> $$
> p(x \mid \theta)
> $$
>
> * 当 \$\theta\$ 固定，\$x\$ 是随机变量 → 是 **密度函数**
> * 当 \$x\$ 固定，\$\theta\$ 是待估参数 → 是 **似然函数**

---

## 2️贝叶斯推断中的角色

### 贝叶斯公式：

$$
p(\theta \mid x) = \frac{p(x \mid \theta) \cdot p(\theta)}{p(x)}
$$

| 项目                 | 名称           | 数学类型              | 意义            |
| ------------------ | ------------ | ----------------- | ------------- |
| $p(\theta)$        | **先验分布**     | 密度函数              | 表达观测数据前对参数的信念 |
| $p(x \mid \theta)$ | **似然函数**     | 似然函数              | 表达参数对数据的解释能力  |
| $p(\theta \mid x)$ | **后验分布**     | 密度函数              | 结合数据后对参数的更新认知 |
| $p(x)$             | **边际似然（证据）** | 常数（相对 \$\theta\$） | 保证后验归一化       |




### $p(x)$ 的具体含义

$p(x)$ 是对参数 $\theta$ 进行边际化后的数据分布，即：

$$
p(x) = \int p(x|\theta) p(\theta) \, d\theta
$$

这里，$p(x)$ 表示在考虑所有可能的参数 $\theta$ 下，数据 $x$ 的整体概率。



### 为什么叫边际分布？

边际分布的名称来源于概率论中的“边际化”操作：

* 联合分布 $p(x, \theta)$ 是两个变量的概率分布。
* 对某个变量积分/求和后，得到另一个变量的边际分布。

例如：

$$
p(x) = \int p(x, \theta) d\theta
$$

这就是“对 $\theta$ 边际化”，所以称为边际分布。



### 作用

* **归一化常数**：保证后验分布 $p(\theta | x)$ 是一个有效的概率分布（积分为1）。
* **模型比较**：$p(x)$ 也称为模型的“证据”，用于计算不同模型的贝叶斯因子，比较模型优劣。




> ✅ 在贝叶斯推断中，**似然函数**用来更新先验分布以获得后验分布。
> ✅ 后验仍然是密度函数，用于后续预测或采样。

---

## 3️图像理解：同一函数两种视角

假设有函数：

$$
p(x \mid \theta) = \mathcal{N}(x; \theta, \sigma^2)
$$

| 视角       | 变量                       | 解释      | 对应函数 |
| -------- | ------------------------ | ------- | ---- |
| **频率派**  | \$x\$ 是变量，\$\theta\$ 是已知 | 用参数解释数据 | 密度函数 |
| **贝叶斯派** | \$x\$ 是已知，\$\theta\$ 是变量 | 用数据估计参数 | 似然函数 |

---

## 4️举个例子

假设观测数据 \$x = 5\$ 来自正态分布：

* 模型：$x \sim \mathcal{N}(\theta, 1)$
* 你想估计 \$\theta\$：

| 内容   | 表达式                                                                                                 | 类型                  |
| ---- | --------------------------------------------------------------------------------------------------- | ------------------- |
| 似然函数 | $L(\theta) = p(x=5 \mid \theta) = \frac{1}{\sqrt{2\pi}} \exp\left(-\frac{(5 - \theta)^2}{2}\right)$ | 以 \$\theta\$ 为变量的函数 |
| 密度函数 | $p(x \mid \theta = 4) = \mathcal{N}(x; 4, 1)$                                                       | 以 \$x\$ 为变量的函数      |

---

## 5️应用对比总结

| 对象       | 类型  | 变量                | 用途       | 典型应用         |
| -------- | --- | ----------------- | -------- | ------------ |
| **密度函数** | PDF | 数据 \$x\$ 是变量      | 给定参数生成数据 | 预测、模拟、采样     |
| **似然函数** | 函数  | 参数 \$\theta\$ 是变量 | 给定数据反推参数 | 最大似然估计、贝叶斯更新 |

---

## 总结口诀

* **密度函数**：看数据怎么来的（\$x\$ 是变量）
* **似然函数**：看哪个参数最好（\$\theta\$ 是变量）
* **贝叶斯推断**：用似然 × 先验 → 得后验




# Q2 频率派 vs 贝叶斯派：原理与比较

---

## 1️核心理念对比（哲学出发点）

| 观点         | 频率派（Frequentist）  | 贝叶斯派（Bayesian）     |
| ---------- | ----------------- | ------------------ |
| **参数的本质**  | **固定未知值**         | **随机变量**（有分布）      |
| **数据的本质**  | 随机变量，服从某种分布       | 同样是随机变量            |
| **不确定性来源** | 来自于数据的采样误差        | 来自于对参数的不确定性（由先验建模） |
| **目标**     | 找到“最可能导致观测数据的参数值” | 更新我们对参数的信念分布       |

简言之：

* **频率派：参数固定，数据随机；只研究采样带来的变化**
* **贝叶斯派：参数也随机；研究我们对参数的主观不确定性**

---

## 2️参数估计方式对比

| 方面         | 频率派                                                            | 贝叶斯派                                                          |
| ---------- | -------------------------------------------------------------- | ------------------------------------------------------------- |
| **主要方法**   | 最大似然估计（MLE）                                                    | 贝叶斯更新（Posterior）                                              |
| **表达式**    | $\hat{\theta}_{\text{MLE}} = \arg\max_\theta p(x \mid \theta)$ | $p(\theta \mid x) = \dfrac{p(x \mid \theta) p(\theta)}{p(x)}$ |
| **是否需要先验** | 否                                                              | 是（主观或经验）                                                      |
| **输出结果**   | 一个点估计（如 \$\hat{\theta} = 0.8\$）                                | 一个后验分布（如 \$\theta \sim \text{Beta}(8,2)\$）                    |

---

## 3️区间推断方式对比

| 问题                            | 频率派                             | 贝叶斯派                                              |
| ----------------------------- | ------------------------------- | ------------------------------------------------- |
| **置信区间（Confidence Interval）** | 一种“采样频率”意义上的区间                  | $P\left(\theta \in [a,b]\right) = 1 - \alpha$ 不成立 |
| **后验区间（Credible Interval）**   | 不使用                             | 直接表示“参数落在该区间的概率为 \$1 - \alpha\$”                  |
| **解释示例**                      | 若重复实验100次，有95次 \$\theta\$ 落在CI内 | 给定数据后，\$\theta\$ 落在该区间的概率为95%                     |

---

## 4️举例说明：硬币抛掷

假设你抛了10次硬币，出现了7次正面。

### （1）频率派推断

* **目标**：估计正面概率 $\theta$
* **方法**：最大似然估计（MLE）

  $$
  \hat{\theta} = \frac{7}{10} = 0.7
  $$
* **解释**：我们假设存在一个固定但未知的正面概率 \$\theta\$，我们只是根据样本来估计它。

### （2）贝叶斯派推断

* **目标**：估计正面概率 $\theta$
* **假设先验**：$\theta \sim \text{Beta}(1,1)$（即均匀先验）
* **更新后验**：

  $$
  \theta \mid x \sim \text{Beta}(8,4)
  $$
* **解释**：我们原来认为\$\theta\$是均匀分布的（任何值都一样可能），观测到7次正面后，我们更新了对 \$\theta\$ 的信念，后验变成 Beta(8,4)。

---

## 5️数学思想差异总结

| 比较维度        | 频率派                       | 贝叶斯派                   |                   |                |                     |
| ----------- |---------------------------|------------------------|-------------------| -------------- |---------------------|
| **估计是哪个量？** | 参数的点估计值                   | 参数的分布                  |                   |                |                     |
| **参数是？**    | 固定常数，未知                   | 随机变量，有先验               |                   |                |                     |
| **核心公式**    | 最大化似然：( $\max _\theta p(x | \theta)$ )             | 贝叶斯更新：( $p(\theta | x) \propto p(x | \theta)p(\theta)$ ) |
| **是否容易过拟合** | 易受数据波动影响                  | 有先验可正则化，有更稳定估计         |                   |                |                     |
| **计算难度**    | 通常更简单                     | 可能需要采样（MCMC）、变分推断等数值方法 |                   |                |                     |

---

## 6️应用场景对比

| 场景          | 推荐方法     | 原因           |
| ----------- | -------- | ------------ |
| 数据充足，先验难以获取 | 频率派方法更高效 | 避免主观先验，易于实现  |
| 数据稀缺，但有领域知识 | 贝叶斯更优    | 可以借助先验知识增强估计 |
| 模型不确定性要量化   | 贝叶斯更好    | 后验分布天然表示不确定性 |
| 实时更新模型      | 贝叶斯天然支持  | 后验可作为下一轮的先验  |

---

## 最后总结口诀

> * **频率派**：参数是“未知但固定”的，数据是“随机的”；不讲先验，追求客观。
> * **贝叶斯派**：参数是“有分布的随机变量”，数据用来更新先验；主观可控，灵活表达不确定性。
